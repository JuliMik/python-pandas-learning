Practicing pandas fundamentals (import, reading CSV, exploring a DataFrame) using a real, large, messy dataset instead of a toy example — the 2025 Stack Overflow Developer Survey (49,191 responses, 172 columns).

## Setup
Import pandas — the core library for working with tabular data in Python.

```python
import pandas as pd
```

## Load the data and the schema (column reference)
Read the raw survey responses into a DataFrame. The survey has 172 columns with non-obvious names — `schema.csv` maps each column code to its full question text.

**New:** using `index_col` to set a meaningful column as the row label instead of the default 0,1,2... — `ResponseId` for the main data, and `qname` (the column code) for the schema, so each row can be looked up directly by its natural identifier.

```python
df = pd.read_csv('data/results.csv', index_col='ResponseId')
schema_df = pd.read_csv('data/schema.csv', index_col='qname')
```

## Display settings
By default, pandas truncates wide/long output with `...`. `set_option` overrides this so all 172 columns and up to 139 rows display in full — useful for exploring a wide dataset like this one.

**Note:** the value and the option name must be two separate arguments — `'display.max_columns', 172` — not one string with the comma inside the quotes (that was an early mistake here, same trap as a misplaced comma in a SQL string literal).

```python
pd.set_option('display.max_columns', 172)
pd.set_option('display.max_rows', 139)
```

## List all column names
Wrapping `df.columns` in `list()` forces pandas to print every column name in full — useful for scanning all 172 question codes before deciding which ones to explore.

```python
list(df.columns)
```

## Explore a single column
`value_counts()` counts how many times each unique value appears in a column — the pandas equivalent of `GROUP BY ... COUNT(*)` in SQL.

```python
df['LearnCodeChoose'].value_counts()
```

Display the schema to see the question text behind each column code.

```python
schema_df
```

## Look up one specific value with `.loc`
`.loc[row_label, column_label]` returns a single value directly — here, the full question text for the row indexed `'ICorPM'`.

**Note:** by default, long text gets truncated with `...`. Fixing this requires a different option than `max_columns` — `pd.set_option('display.max_colwidth', None)` removes the character limit per cell (not shown here to avoid overriding useful truncation elsewhere, but useful to know for reading full question text).

```python
schema_df.loc['ICorPM', 'question']
```

## Sort by index
`sort_index()` sorts the DataFrame by its row labels (here, the `qname` codes) rather than by a column's values. `ascending=False` sorts Z→A instead of the default A→Z.

```python
schema_df.sort_index(ascending=False)
```
