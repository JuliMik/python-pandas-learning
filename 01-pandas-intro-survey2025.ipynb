## Display settings
By default, pandas truncates wide/long output with `...`. `set_option` overrides this so all 172 columns and up to 139 rows display in full — useful for exploring a wide dataset like this one.

**Note:** the value and the option name must be two separate arguments — `'display.max_columns', 172` — not one string with the comma inside the quotes (that was an early mistake here, same trap as a misplaced comma in a SQL string literal).

```python
pd.set_option('display.max_columns', 172)
pd.set_option('display.max_rows', 139)
```

## List all column names
`df.columns` alone gets truncated with `...` for wide DataFrames. Wrapping it in `list()` forces pandas to print every column name in full — useful for scanning all 172 question codes before deciding which ones to explore.

```python
list(df.columns)
```

## Explore a single column
`value_counts()` counts how many times each unique value appears in a column — the pandas equivalent of `GROUP BY ... COUNT(*)` in SQL. Here it shows how respondents answered the "how do you prefer to learn to code" question.

```python
df['LearnCodeChoose'].value_counts()
```

## Slice specific rows and columns with `.loc`
`.loc[rows, columns]` selects a sub-section of the DataFrame by label. Here: rows 0 to 2 (inclusive), and every column from `LearnCodeChoose` to `AILearnHow` — a way to zoom into a specific block of related survey questions without scrolling through all 172 columns.

```python
df.loc[0:2, 'LearnCodeChoose':'AILearnHow']
```
