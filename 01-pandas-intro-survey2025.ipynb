## Display settings
By default, pandas truncates wide/long output with `...`. `set_option` overrides this so all 172 columns and up to 139 rows display in full — useful for exploring a wide dataset like this one.

**Note:** the value and the option name must be two separate arguments — `'display.max_columns', 172` — not one string with the comma inside the quotes (that was an early mistake here, same trap as a misplaced comma in a SQL string literal).

```python
pd.set_option('display.max_columns', 172)
pd.set_option('display.max_rows', 139)
```
